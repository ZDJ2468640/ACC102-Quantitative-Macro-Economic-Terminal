```python
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# 1. 模拟宏观经济数据集 (2010-2023)
countries = ['China', 'United States', 'United Kingdom', 'Japan', 'Germany']
years = list(range(2010, 2024))
data_list = []

for country in countries:
    for year in years:
        inf = np.random.uniform(1.0, 5.0) if country != 'Japan' else np.random.uniform(-0.5, 1.5)
        rir = np.random.uniform(0.5, 3.0) - (inf * 0.6) # 模拟政策关联性
        data_list.append({'Country': country, 'Year': year, 'Inflation': round(inf, 2), 'Real_Interest_Rate': round(rir, 2)})

df = pd.DataFrame(data_list)

# 2. 机器学习预演 (Linear Regression) - 展示深度
X = df[['Inflation']]
y = df['Real_Interest_Rate']
model = LinearRegression().fit(X, y)
print(f"Global Sensitivity Coefficient: {model.coef_[0]:.4f}")

# 3. 保存为本地CSV (确保相对路径)
df.to_csv('macro_data.csv', index=False)
print("✅ macro_data.csv generated successfully!")
```
